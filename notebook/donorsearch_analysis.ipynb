```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from dotenv import load_dotenv
import os
from sqlalchemy import create_engine

load_dotenv()

engine = create_engine(
    f"postgresql://{os.getenv('DB_USER')}:{os.getenv('DB_PASS')}@"
    f"{os.getenv('DB_HOST')}:{os.getenv('DB_PORT', '5432')}/{os.getenv('DB_NAME')}"
)
```

```python
user_anon_data = pd.read_sql("SELECT * FROM donorsearch.user_anon_data", engine)
donation_anon = pd.read_sql("SELECT * FROM donorsearch.donation_anon", engine)
donation_plan = pd.read_sql("SELECT * FROM donorsearch.donation_plan", engine)
user_anon_bonus = pd.read_sql("SELECT * FROM donorsearch.user_anon_bonus", engine)
user_donation = pd.read_sql("SELECT * FROM donorsearch.user_donation", engine)
events = pd.read_sql("SELECT * FROM donorsearch.events", engine)
bs_data = pd.read_sql("SELECT * FROM donorsearch.bs_data", engine)
```
